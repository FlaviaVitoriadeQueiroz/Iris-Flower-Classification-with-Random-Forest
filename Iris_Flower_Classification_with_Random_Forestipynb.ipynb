ATIVIDADE - COMBINAÇÃO DE CLASSIFICADORES

Técnica utilizada: Random Forest

Biblioteca: scikit-learn


Importação das bibliotecas

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix)
import matplotlib.pyplot as plt
import seaborn as sns
```

CARREGAMENTO DO DATASET

```python
iris = load_iris()
```

Variáveis de entrada

```python
X = iris.data
```

Variável alvo

```python
y = iris.target
```

DIVISÃO DOS DADOS

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)
```

CRIAÇÃO DO MODELO

```python
modelo = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

Treinamento

```python
modelo.fit(X_train, y_train)
```

Previsões

```python
y_pred = modelo.predict(X_test)
```

AVALIAÇÃO DO MODELO

Acurácia

```python
acuracia = accuracy_score(y_test, y_pred)

print("Acurácia do modelo:", acuracia)
```

Relatório de classificação

```python
print("\nRelatório de Classificação:")
print(classification_report(y_test, y_pred))
```

Matriz de confusão

```python
cm = confusion_matrix(y_test, y_pred)

print("\nMatriz de Confusão:")
print(cm)
```

MATRIZ DE CONFUSÃO VISUAL

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.title('Confusion Matrix - Random Forest')
plt.xlabel('Predicted Class')
plt.ylabel('Actual Class')

plt.show()
```

Análise dos Resultados

Após o treinamento do modelo Random Forest, foram realizadas previsões utilizando o conjunto de teste e avaliadas por meio de métricas de classificação.

A principal métrica analisada foi a acurácia, que representa a proporção de classificações corretas realizadas pelo modelo. Os resultados obtidos demonstraram um elevado nível de desempenho, indicando que o algoritmo foi capaz de identificar corretamente a maioria das espécies presentes no conjunto de dados.

O relatório de classificação permitiu analisar individualmente cada classe por meio das métricas de Precision, Recall e F1-Score. Os valores obtidos foram elevados para todas as espécies, evidenciando que o modelo apresentou boa capacidade tanto para identificar corretamente os exemplos positivos quanto para minimizar erros de classificação.

A matriz de confusão confirmou esse comportamento ao apresentar uma predominância de valores na diagonal principal, que representam as classificações corretas. A baixa ocorrência de valores fora da diagonal indica que poucas amostras foram classificadas incorretamente.

Esse desempenho pode ser explicado pelas características do dataset Iris, que apresenta atributos altamente discriminativos entre as espécies, especialmente para a espécie Setosa. Além disso, a utilização do algoritmo Random Forest contribuiu para aumentar a robustez das previsões ao combinar múltiplas árvores de decisão em um único modelo.

De maneira geral, os resultados demonstram que o modelo foi capaz de aprender os padrões presentes nos dados e realizar classificações com elevada precisão. A combinação de classificadores utilizada pelo Random Forest mostrou-se eficiente para o problema proposto, produzindo um modelo estável e com boa capacidade de generalização.

Análise do Código

O código implementa um modelo de classificação supervisionada utilizando o algoritmo Random Forest para identificar espécies de flores do conjunto de dados Iris.

Inicialmente, são importadas as bibliotecas necessárias para carregamento dos dados, divisão dos conjuntos de treinamento e teste, construção do modelo e avaliação dos resultados. Também são utilizadas bibliotecas de visualização para representar a matriz de confusão de forma gráfica.

Na etapa de carregamento dos dados, é utilizado o dataset Iris disponibilizado pela biblioteca Scikit-Learn. Esse conjunto de dados contém informações sobre três espécies de flores (Setosa, Versicolor e Virginica) e quatro características morfológicas: comprimento e largura da sépala, além de comprimento e largura da pétala.

As variáveis de entrada (X) são compostas pelas medidas das flores, enquanto a variável alvo (y) corresponde à espécie de cada flor.

Em seguida, os dados são divididos em conjuntos de treinamento e teste utilizando a função train_test_split(). Nesse processo, 70% dos dados são destinados ao treinamento do modelo e 30% são reservados para avaliação. O parâmetro random_state=42 garante a reprodutibilidade dos resultados.

O algoritmo escolhido foi o Random Forest, uma técnica de Ensemble Learning baseada na combinação de múltiplas Árvores de Decisão. Cada árvore é treinada utilizando subconjuntos diferentes dos dados e das características disponíveis. Durante a predição, as árvores realizam uma votação e a classe mais votada é escolhida como resultado final. Essa abordagem reduz a variância do modelo e diminui o risco de overfitting.

Após a criação do modelo, é realizado o treinamento utilizando os dados de treinamento por meio do método fit(). Em seguida, o método predict() é utilizado para gerar previsões sobre o conjunto de teste.

A avaliação do desempenho é realizada por meio de três métricas principais. A primeira é a acurácia, que mede a proporção de previsões corretas em relação ao total de amostras avaliadas. A segunda é o Classification Report, que apresenta métricas como precisão (Precision), revocação (Recall) e F1-Score para cada uma das espécies analisadas. Por fim, é utilizada a matriz de confusão, que permite visualizar detalhadamente os acertos e erros de classificação para cada classe.

Além da matriz de confusão textual, o projeto inclui uma representação gráfica utilizando Seaborn e Matplotlib. Essa visualização facilita a interpretação dos resultados, permitindo identificar rapidamente quais espécies foram corretamente classificadas e quais apresentaram confusão durante o processo de predição.

De forma geral, o código apresenta uma implementação adequada de um problema de classificação supervisionada, contemplando todas as etapas fundamentais de um projeto de Machine Learning: carregamento dos dados, preparação, treinamento, predição, avaliação e visualização dos resultados. A utilização do Random Forest demonstra o emprego de técnicas de combinação de classificadores, capazes de produzir modelos mais robustos e com melhor capacidade de generalização.
